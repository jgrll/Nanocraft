# View your structures
You can use the following built-in methods to view all your structures easily.
- `visualize_xyz`: view all structures provided in a list. Use a py3Dmol viewer
- `visualize_nc_core`: view the core of provided structures. Same viewer method as `visualize_xyz`.
- `visualize_nc_surf`: view the outer layer of the bulk core of provided structures. Same viewer method as `visualize_xyz`.
- `visualize_nc_ligands`: view the all detected ligands atoms from provided structures. Same viewer method as `visualize_xyz`.
- `visualize_nc_all`: calls successively all 3 `visualize_nc_*` side by side for easier analysis. Same viewer method as `visualize_xyz`.
- `visualize_facets`: view all atoms belonging to facet(s), edge(s) of vertex(ices). **Works only on bare structures**. Same viewer method as `visualize_xyz`.
- `visualize_simplices`: View all simplices generated from the tessellation algorithm. Can be usefull to choose a value for the alpha parameter. **Works only on bare structures**. Use plotly for the representation of polygons.
- `visualize_patches`: Same as `visualize_simplices`, but regroup the simplices with almost colinear normal vectors into polygons of higher order. **Works only on bare structures**.
- `visualize_local_frame`: Same method as `visualize_xyz`, but allows to add a local frame (you need to provide the origin and axis vectors).
